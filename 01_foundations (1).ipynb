This notebook covers fundamental Python concepts applied to the **Garissa 30min Modeling Dataset**, including:

6. Two small custom functions relevant to solar energy analysis

## 1. Variables
Variables are used to store data values. Below, we define basic parameters for our analysis, such as dataset file paths and threshold values.

```python
# Defining variables
dataset_filename = "Garissa 30min Modeling Dataset.xlsx"
site_name = "Garissa Solar Plant"
temp_threshold = 35.0  # Celsius threshold for high cell temperature

print(f"Analyzing site: {site_name}")
print(f"Target file: {dataset_filename}")
print(f"High cell temperature threshold: {temp_threshold} °C")
```

## 2. Lists
Lists store multiple items in a single variable. Here, we list key meteorological and performance variables we want to explore from our dataset.

```python
# Creating a list of key column names to investigate
key_columns = [
    "Energy Yield",
    "Environment Temperature",
    "Wind speed",
    "POA global",
    "Tcell"
]

print("Columns to investigate:")
for col in key_columns:
    print(f"- {col}")
```

## 3. Dictionaries
Dictionaries store data in key-value pairs. Below, we map dataset metadata and summary statistics.

```python
# Creating a dictionary for site metadata
site_metadata = {
    "location": "Garissa",
    "latitude": -0.4532,
    "longitude": 39.6460,
    "resolution_minutes": 30,
    "total_records": 14846
}

print("Site Metadata Dictionary:")
for key, value in site_metadata.items():
    print(f"{key.capitalize()}: {value}")
```

## 4. Loops
Loops allow us to iterate over sequences. Here, we load our dataset using Pandas and use a `for` loop to inspect basic statistics for each key column in our list.

```python
import pandas as pd

# Load the dataset
df = pd.read_excel(dataset_filename, sheet_name='Data')

# Using a loop to calculate and print mean values for each key column
print("Mean values of key columns:")
for col in key_columns:
    mean_val = df[col].mean()
    print(f"{col}: {mean_val:.2f}")
```

## 5. Conditionals
Conditionals (`if`, `elif`, `else`) make decisions in code. Here, we check the average cell temperature against our threshold variable to categorize thermal conditions.

```python
# Conditional check on average cell temperature
avg_tcell = df['Tcell'].mean()

print(f"Overall Average Cell Temperature: {avg_tcell:.2f} °C")

if avg_tcell > temp_threshold:
    print("Status: High thermal operating environment (Average Tcell exceeds threshold).")
elif avg_tcell > 30.0:
    print(
        "Status: Moderate thermal operating environment (Average Tcell is between 30°C and threshold).")
else:
    print("Status: Normal thermal operating environment.")
```

## 6. Functions Relevant to the Investigation
We define two custom functions:
1. `calculate_power_ratio`: Computes the ratio of Energy Yield to Plane of Array (POA) global irradiance.
2. `categorize_wind_speed`: Classifies wind speed into qualitative categories to assess cooling effects.

```python
# Function 1: Calculate energy yield per unit of POA global irradiance
def calculate_power_ratio(energy_yield, poa_global):
    """Calculates ratio of energy yield to POA global irradiance, handling zero division."""
    if poa_global == 0 or pd.isna(poa_global) or pd.isna(energy_yield):
        return 0.0
    return energy_yield / poa_global

# Function 2: Categorize wind speed for thermal analysis
def categorize_wind_speed(wind_speed):
    """Categorizes wind speed into Calm, Moderate, or Windy."""
    if pd.isna(wind_speed):
        return "Unknown"
    elif wind_speed < 1.5:
        return "Calm"
    elif wind_speed < 4.0:
        return "Moderate"
    else:
        return "Windy"

# Test our functions on the first few rows of the dataset
df['Sample_Power_Ratio'] = df.apply(lambda row: calculate_power_ratio(row['Energy Yield'], row['POA global']), axis=1)
df['Wind_Category'] = df['Wind speed'].apply(categorize_wind_speed)

print(df[['DateTime', 'Energy Yield', 'POA global', 'Sample_Power_Ratio', 'Wind speed', 'Wind_Category']].head(10))
```
